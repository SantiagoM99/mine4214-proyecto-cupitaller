## 00. Exploración inicial

Lista los libros de Excel de Bronze (`data/bronce/Bookeau/` si están los datos reales; si no, `data/bronce/dummy/`) y resume el inventario por grupo. No modifica los datos.

El enunciado y el plan están en `docs/documentacion/`.

**Antes de ejecutar:** desde la raíz del proyecto, correr `python3 src/main.py --dummy` (datos ficticios versionados) o `python3 src/main.py` (datos reales, que no se versionan). El cuaderno solo lee los resultados del pipeline en `docs/artifacts/`, `data/plata/` y `data/oro/`. Con datos ficticios los conteos son pequeños y no coinciden con las cifras del informe, que usan los datos reales.


```python
from pathlib import Path

candidatos = [Path.cwd(), *Path.cwd().parents]
PROYECTO = next(
    (ruta for ruta in candidatos
     if (ruta / "src" / "main.py").is_file()),
    None,
)
if PROYECTO is None:
    raise RuntimeError("Abre el cuaderno desde la carpeta Proyecto o notebooks.")
# Datos reales si están disponibles; si no, los ficticios versionados en data/bronce/dummy/.
REALES = PROYECTO / "data" / "bronce" / "Bookeau"
DATOS = REALES if any(REALES.rglob("*.xlsx")) else PROYECTO / "data" / "bronce" / "dummy"
print("Proyecto:", PROYECTO)
print("Fuentes:", DATOS)
```

```python
archivos = sorted(
    ruta for ruta in DATOS.rglob("*")
    if ruta.is_file() and not ruta.name.startswith(".")
)
if not archivos:
    print("No hay archivos en", DATOS)
else:
    for ruta in archivos:
        print(f"{ruta.relative_to(DATOS)} | {ruta.suffix or 'sin extensión'} | {ruta.stat().st_size:,} bytes")
```

```python
import csv
from collections import defaultdict

INVENTARIO = PROYECTO / "docs" / "artifacts" / "caracterizacion" / "archivos.csv"
if not INVENTARIO.is_file():
    raise RuntimeError("Ejecuta primero desde la raíz: python3 src/main.py --dummy")
with INVENTARIO.open(newline="", encoding="utf-8") as archivo:
    inventario = list(csv.DictReader(archivo))

grupos = defaultdict(list)
for fuente in inventario:
    grupos[fuente["grupo"]].append(fuente)

for nombre, fuentes in sorted(grupos.items()):
    archivos = len({fuente["archivo"] for fuente in fuentes})
    filas = sum(int(fuente["filas"]) for fuente in fuentes)
    print(f"{nombre}: {archivos} archivos, {filas:,} filas de datos")

print("Estos conteos aún no representan reservas o estudiantes únicos.")
```
