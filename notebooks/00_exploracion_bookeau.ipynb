## 00. Exploración inicial

Fuentes originales extraídas en `data/bronce/Bookeau/`; ZIP conservado en `data/bronce/Bookeau.zip`.

El enunciado está en `docs/Proyecto - entrega 1.pdf` y los requisitos en `docs/plan_entrega_1.md`. El inventario inicial está en `docs/inventario_fuentes.csv`.

Objetivo: revisar el inventario antes de perfilar las fuentes y definir los dos análisis. Este cuaderno no modifica los datos originales.


```python
from pathlib import Path

candidatos = [Path.cwd(), *Path.cwd().parents]
PROYECTO = next(
    (ruta for ruta in candidatos
     if (ruta / "docs" / "plan_entrega_1.md").is_file()),
    None,
)
if PROYECTO is None:
    raise RuntimeError("Abre el cuaderno desde la carpeta Proyecto o notebooks.")
DATOS = PROYECTO / "data" / "bronce" / "Bookeau"
print("Proyecto:", PROYECTO)
print("Fuentes originales:", DATOS)
```

```python
archivos = sorted(
    ruta for ruta in DATOS.rglob("*")
    if ruta.is_file() and not ruta.name.startswith(".")
)
if not archivos:
    print("Pendiente: copiar los archivos de Bookeau en data/bronce/.")
else:
    for ruta in archivos:
        print(f"{ruta.relative_to(DATOS)} | {ruta.suffix or 'sin extensión'} | {ruta.stat().st_size:,} bytes")
```

```python
import csv
from collections import defaultdict

with (PROYECTO / "docs" / "inventario_fuentes.csv").open(newline="") as archivo:
    inventario = list(csv.DictReader(archivo))

grupos = defaultdict(list)
for fuente in inventario:
    grupos[fuente["grupo"]].append(fuente)

for nombre, fuentes in sorted(grupos.items()):
    archivos = len({fuente["archivo"] for fuente in fuentes})
    filas = sum(int(fuente["filas_datos"]) for fuente in fuentes)
    print(f"{nombre}: {archivos} archivos, {filas:,} filas de datos")

print("Estos conteos aún no representan reservas o estudiantes únicos.")
```
